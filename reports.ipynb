### Problem Statement and Initial Data Loading :
This section establishes the foundational context for the project: addressing the critical need for automated suspicious transaction detection in financial services to combat illicit activities like money laundering. The core objective is to develop an unsupervised machine learning system capable of identifying anomalous patterns in transaction data. The practical implementation begins with loading the Trx_ML.csv dataset into a pandas DataFrame (transactions_df). An initial review of the DataFrame's dimensions and column names is performed to ensure data integrity and prepare for subsequent processing. This step is crucial for confirming the availability and basic structure of the raw data required for the entire anomaly detection pipeline

###Data Preparation and Feature Engineering

This comprehensive report details the essential data preparation and preprocessing steps, which are vital for transforming the raw transaction data into a clean, normalized, and feature-rich format suitable for machine learning models. The process involves:

**Date Feature Extraction**: The CreatedOn column is meticulously converted to datetime objects, enabling the extraction of CreatedOn_hour — a temporal feature crucial for identifying time-based anomalies.

**Column Refinement:** Irrelevant columns, specifically ChequeDate, are systematically removed to reduce noise and focus the dataset on pertinent information.

**Feature Selection for Anomaly Detection**: Key features are deliberately chosen based on their potential to highlight suspicious activity. These include numerical attributes (LocalAmount, CreatedOn_hour) and categorical identifiers (Anon_AccountId, TransactionChannels).

**Robust Missing Value Handling**: Missing numerical values are addressed by dropping the corresponding rows, while missing categorical entries are imputed with 'Unknown' to maintain data completeness and prevent model errors.
**Feature Transformation and Scaling:**

 A ColumnTransformer orchestrates the application of distinct preprocessing techniques:
StandardScaler is applied to numerical features, ensuring they have a mean of zero and unit variance, which is critical for distance-based algorithms.
OneHotEncoder transforms categorical features into a binary vector representation, making them amenable to machine learning algorithms while handling unseen categories gracefully.
The output, X_scaled_unsupervised, represents the fully processed dataset, poised for the application of unsupervised anomaly detection algorithms.



 ### DBSCAN Anomaly Detection Model Implementation
This report describes the application of DBSCAN (Density-Based Spatial Clustering of Applications with Noise) as another unsupervised anomaly detection method. The implementation details are as follows:

**Data Subset Selection**: To manage computational

 complexity while demonstrating the algorithm, a subset consisting of the last 100,000 rows from the original transactions_df is utilized.

**Preprocessing of Subset:** The previously trained

preprocessor is applied to this subset, ensuring that the input features to DBSCAN are consistently scaled and encoded.

**DBSCAN Model Configuration**: A DBSCAN model is

 initialized with key parameters: eps=0.5 (the maximum distance between two samples for one to be considered as in the neighborhood of the other) and min_samples=10 (the minimum number of samples in a neighborhood for a point to be considered a core point). These parameters are crucial for defining cluster density and identifying noise.

**Clustering and Anomaly Identification:** The fit_predict

 method is invoked, which performs the clustering and assigns a cluster label to each data point. Notably, points that do not belong to any dense cluster are labeled as -1, signifying anomalies or 'noise' points.

**Result Integration and Analysis:** The DBSCAN cluster labels are added to the transactions_df_subset as DBSCAN_Cluster. An additional column, Is_Anomaly_DBSCAN, is created to explicitly mark anomalies (-1) or normal points (1), further mapped to human-readable labels (Suspicious, Normal).

**Quantitative and Qualitative Outpu**t: The report

 highlights that DBSCAN identified 37493 suspicious transactions within the subset. The head of the subset DataFrame with anomaly labels and specific examples of detected suspicious transactions are displayed. Furthermore, a value count of DBSCAN_Cluster reveals the distribution of formed clusters and the total count of identified noise points.

### K-Means Anomaly Detection Model Implementation

This report outlines the application of K-Means clustering as an unsupervised anomaly detection technique. The process involves:

**Data Subset Selection:** A subset consisting of the last 300,000 rows of the transactions_df is chosen to focus the K-Means training on a manageable yet representative sample of recent transactions.

**Preprocessing Application:** The pre-fitted preprocessor (from the data preparation stage) is applied to this subset, ensuring consistency in scaling and encoding of features.

**K-Means Model Training:** A KMeans model is initialized with n_clusters=5 and random_state=42 for reproducibility. The model is then fitted to the scaled data, assigning each data point to one of the five clusters.

**Anomaly Scoring with Distance Metrics:** The core of K-Means anomaly detection lies in calculating the Euclidean distance of each data point to its assigned cluster centroid. Data points located far from their cluster centers are considered potential anomalies.

**Anomaly Thresholding:** An anomaly threshold is determined by calculating the 99th percentile of these distances. Transactions whose distances exceed this threshold are classified as 'Suspicious'
.
**Result Integration and Analysis: **The cluster assignments, distances to centroids, and final anomaly labels are integrated back into the transactions_df_subset_kmeans DataFrame. The model successfully identified 3000 transactions as suspicious. A distribution of cluster counts is also provided.

**Model Persistence:** The trained K-Means model, along with its preprocessor, feature lists, and the calculated threshold, are serialized and saved to kmeans_model.pkl. This ensures the model can be loaded and reused consistently for future anomaly scoring and deployment without retraining.


###Model Deployment via Streamlit Application
This report focuses on the deployment aspect of the anomaly detection solution through a user-friendly Streamlit web application (app.py). This application provides an interactive interface for end-users to leverage the trained anomaly detection models. Key functionalities and components of the deployment include:

**Dynamic Model Loading:** The Streamlit app is designed to dynamically load either the dbscan_model.pkl or kmeans_model.pkl based on user selection. This loading process also retrieves the associated preprocessor, numerical_features, categorical_features, and, for K-Means, the threshold for consistent inference.

**User Interface for Model Selection:** A dropdown menu allows users to choose between the deployed DBSCAN and K-Means models, demonstrating flexibility in model application.

**Single Transaction Prediction:** The application features an intuitive form where users can input details for a single transaction. The selected anomaly detection model then processes these inputs and provides an immediate prediction (Suspicious/Normal), with K-Means predictions based on distance to centroids relative to a threshold, and DBSCAN predictions based on cluster assignment.

**Batch Anomaly Detection via CSV Upload:** A crucial feature is the ability for users to upload a CSV file containing multiple transactions. The app automatically preprocesses this batch data, applies the chosen model, and displays the results in a DataFrame, with suspicious transactions visually highlighted.

**Robust Error Handling:** The application includes error and warning messages to guide users, for instance, if a model file is not found or if the uploaded data lacks required features.
This Streamlit deployment provides a practical, accessible, and interactive way for stakeholders to interact with the anomaly detection system, perform real-time checks, and analyze batch data, thereby transforming the machine learning models into actionable tools.
